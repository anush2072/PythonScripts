```python
import os
import boto3
profile_name = 'my_profile_name'  # replace with your AWS profile name

# run the "aws sso login" command to refresh your AWS credentials
os.system(f'aws sso login --profile {profile_name}')
boto3.setup_default_session(profile_name=profile_name)
s3 = boto3.resource('s3')
for bucket in s3.buckets.all():
  print(bucket.name)

```
